# Plotting Outliers
## using Box-plot, String-plot, Voilon-plot, etc

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/Used_vehicle_data.csv')
```

```python
display(df['flatrate'])
```

```python
display(df['flatrate'].describe())
```

```python
# df.boxplot(df['flatrate'])
df['SELLING_PRICE'].plot.box()
plt.show()
```

```python
# Reload the original dataset to restore the 'flatrate' column
vehicle_df = pd.read_csv('/content/Used_vehicle_data.csv')

# Create a scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(vehicle_df.index, vehicle_df['SELLING_PRICE'], alpha=0.6, color='blue', edgecolors='w')
plt.title('Scatter Plot of SELLING_PRICE')
plt.xlabel('Index')
plt.ylabel('SELLING_PRICE Value')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import seaborn as sns

# Create a violin plot for flatrate
plt.figure(figsize=(8, 6))
sns.violinplot(y=vehicle_df['SELLING_PRICE'], color='lightblue', inner='quartile')
plt.title('Violin Plot of SELLING_PRICE')
plt.ylabel('SELLING_PRICE Value')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Identify rows with extreme outliers in 'flatrate' (values above 50,000)
extreme_outliers = vehicle_df[vehicle_df['SELLING_PRICE'] > 50000]
display(extreme_outliers)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Subplot 1: Strip Plot without Jitter
sns.stripplot(ax=axes[0], y=vehicle_df['SELLING_PRICE'], jitter=False, color='crimson', size=6, alpha=0.7)
axes[0].set_title('Strip Plot of SELLING_PRICE (No Jitter)')
axes[0].set_ylabel('SELLING_PRICE Value')
axes[0].grid(True, linestyle='--', alpha=0.5)

# Subplot 2: Jitter Plot (Strip Plot with Jitter)
sns.stripplot(ax=axes[1], y=vehicle_df['SELLING_PRICE'], jitter=True, color='teal', size=6, alpha=0.7)
axes[1].set_title('Jitter Plot of SELLING_PRICE (With Jitter)')
axes[1].set_ylabel('SELLING_PRICE Value')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Create a Quantile-Box Plot (Boxenplot/Letter-Value Plot)
plt.figure(figsize=(10, 6))
sns.boxenplot(y=vehicle_df['SELLING_PRICE'], color='mediumpurple')
plt.title('Quantile-Box Plot (Boxenplot) of SELLING_PRICE')
plt.ylabel('SELLING_PRICE Value')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a histogram with a rug plot
plt.figure(figsize=(12, 6))
sns.histplot(data=vehicle_df, x='SELLING_PRICE', bins=50, kde=True, color='teal', alpha=0.6)
sns.rugplot(data=vehicle_df, x='SELLING_PRICE', color='crimson', height=0.05)

plt.title('Histogram of SELLING_PRICE with Rug Plot')
plt.xlabel('SELLING_PRICE Value')
plt.ylabel('Count')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
